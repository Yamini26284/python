### **Comments**
---


**single line comment**  
comments are either single line comments which is hash(#).

You do not have to type these manually. In code editors like VS Code, you can highlight a block of text and press Ctrl + / (or Cmd + / on Mac) to comment out multiple lines instantly.  

**Python does not have a native multi-line comment syntax**  


**Docstrings (Documentation Strings)**

If you place that exact same triple-quoted string as the very first line inside a function, a class, or a module, it officially becomes a Docstring.
python

    def add_numbers(a, b):
        """
        This is officially a Docstring.
        It explains what this specific function does.
        """
        return a + b

• *Why it's unique*: Python does not ignore docstrings. It actively binds the text to that function as metadata. Other programmers (or documentation tools) can look up what your function does later.


**Unassigned Triple-Quoted Strings**

If you place a triple-quoted string (""" or ''') randomly in your code without assigning it to a variable, Python will read it as a string literal, realize nothing is using it, and immediately discard it.

    """
    This is technically a multi-line string literal.
    Because it isn't assigned to a variable, Python discards it.
    People use this as a hack for multi-line comments.
    """
    print("Hello World")

• *Why it's not a true comment*: A true # comment leaves absolutely nothing behind when your code is compiled into bytecode. A triple-quoted string is still temporarily read as data by the computer before being tossed away.


```python
""" DocString example """
def welcome_user(username):
    """
    This is officially a Docstring.
    It explains that this function greets a user by their name.
    Python keeps this text saved inside the function's memory!
    """
    return f"Hello, {username}! Welcome back."

# 1. Run the code normally
print("--- Running Function ---")
message = welcome_user("Yamini")
print(message)

print("\n--- Reading the Saved Docstring ---")
# 2. Ask Python to print the documentation hidden inside the function
print(welcome_user.__doc__)
```

```python
# triple-quote string
print("--- Program Started ---")

"""
This is an unassigned triple-quoted string.
It is sitting completely out in the open workspace.
Python will read this text, do absolutely nothing with it, 
and throw it right into the trash.
This makes it act exactly like a multi-line comment block!
"""

print("--- Program Ended ---")
print("The triple-quoted block above ran silently without any errors.")
```

#### If a triple quote is used inside a function block, it is completely valid and saved—but only if it is the very first line

```python
'''
If the triple quote is the very first line directly under the function definition, Python recognizes it as an official Docstring. It stays inside the computer's memory, attached directly to that function.
'''
def active_function():
    """THIS IS SAVED!
    Because it is the very first line inside the block,
    Python saves it in memory as metadata.
    """
    print("Hello")
```
