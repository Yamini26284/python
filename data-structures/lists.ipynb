```python
# traversing on list : running a loop on list

r = [12,23,34,45,56,67,78,9,3]

print(r) # traversing on list

# traversnig on values
for i in r:
    print(i)
    
# traversnig on index
for i in range(0,len(r)): # to reach the end of list wirte len(list)
    print(f"At inedx value {i} we have value {r[i]}")
```

```python
# Questions

# print all positives and negatives seperately in a list.
positives = []
negatives = []
l = [3,-1,4,-5,9]
for i in l: # need to access all the elements
    if i >= 1:
        positives.append(i)
    elif i< 1 :
        negatives.append(i)
    else :
        print(i,"is not a valid number")
print(positives,negatives)
```

```python
# find the mean(avg) of all elements
l = [10,20,30,40]
sum = 0
for i in l:
    sum += i
Mean = sum / len(l)
print(Mean)
```

```python
# find greatest elements and print its index
l = [10,20,50,30,40,51]
x = 0
for i in range(1,len(l)):
    if l[i] > l[x]:
        x = i
print(l[x],i)
```
