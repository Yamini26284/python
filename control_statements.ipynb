```python
# century year : a year that is a multiple of 100, like 2000,1900. so for them divide by 400 to get a leap year
# normal year : just divide it by 4

'''Here are the strict rules narrowed down to three simple conditions:
1. Rule 1 (The Basic Rule): Every year that is evenly divisible by 4 is a candidate to be a leap year.
2. Rule 2 (The Exception): If that year is also evenly divisible by 100, its leap year status is stripped away.
3. Rule 3 (The Overrule): If that year is further evenly divisible by 400, its leap year status is given back.'''


year = int(input("enter the year : "))
if (year % 4 == 0 and year %100 != 0) or (year % 400 ==0) :
    
    print(f"{year} is a leap year")
else:
    print("not a leap year")
    
'''Think of it like two completely separate doors to get into a club:
• Door 1 (Normal Years): Must be divisible by 4 AND NOT a century year.
• Door 2 (Century Years): Must be divisible by 400.
A year only needs to pass through either Door 1 OR Door 2 to be a leap year.'''
```
