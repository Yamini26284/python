### Unicode
Unicode in Python is a system that allows your code to understand, process, and display every character from every language in the world, including emojis and special symbols.

Think of Unicode as a giant, universal dictionary. In this dictionary, every single letter, symbol, or emoji is assigned a unique code number called a code point. For example, the English letter "A" is number 65, the Euro symbol "€" is number 8364, and the laughing emoji "😂" is number 128514.

In Python 3, **every regular string you type is automatically a Unicode string**. You don't have to do anything special to use it


The Two Useful Commands: ord() and chr()

Python gives you two built-in tools to look up things in this universal dictionary:

    1.  ord(character): Give it a character, and it tells you its Unicode number.  
    2.  chr(number): Give it a Unicode number, and it tells you the character.

```python
a = "a" # ord expects a character
print(ord(a))
```

```python
r = 89
print (chr(r))
```

Python string slicing uses the format x[start:stop:step].  

Think of strings as a row of boxes where index positions sit right on the boundaries between letters.

Positive indexes count forward from the left starting at 0, while negative indexes count backward from the right starting at -1.  

 The most critical rule to remember is that Python always excludes the stop index. It walks up to that boundary and stops exactly one step before it, never including the character sitting at the stop position itself.  

When you write x[-1:0:-1], you tell Python to start at the very last letter and step backward by -1. Because of the exclusion rule, when it approaches index 0, it halts right before reaching it. This leaves out the very first letter of the string. 
To fix this and reverse a string completely, you use the shortcut x[::-1]. By leaving the start and stop positions blank, Python dynamically adjusts to your step direction. Since walking backward is impossible from the front of the word, Python automatically flips its defaults, starting at the very end and processing the entire string all the way to the absolute beginning.

Conversely, when you leave the step positive or blank, Python moves forward. A slice like x[:-1] tells Python to start at the beginning and move forward normally, but to stop exactly one step before index -1. This effectively chops off the very last letter of the word. 

Slicing gives you total control over text data simply by shifting these boundary numbers and changing your walking direction.

```python
# Implicit type conversion

''' implicit type conversion (also called coercion) in Python:
• Automatic Promotion: Python automatically upgrades narrower data types to wider data types to prevent data loss (e.g., adding an integer to a float automatically results in a float).
• Float Dominance: Mixing an int and a float in any math operation always outputs a float (e.g., 5 + 2.0 becomes 7.0).
• Boolean Evaluation: Booleans are treated as integers in arithmetic; True converts to 1 and False converts to 0 (e.g., True + 5 becomes 6).
• Strict Type Safety: Python blocks implicit conversions between fundamentally incompatible types, meaning it will never automatically convert a string to a number (e.g., "Age: " + 25 throws a TypeError).
'''

a = 12
print(a/2)
```
