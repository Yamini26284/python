```python
# break : coming out of the control block or stop the loop 
# continue : skipping that specific iteration
# else(works with break) : 

for i in range(1,11):
    if i == 4:
        continue
    print(i)
    
for i in range(1,11):
    if i == 4:
        break
    print(i)
```

```python
'''
In Python, you can pair an else block directly with a for or while loop.
The rule for how it works is simple: The else block runs ONLY if the loop finishes completely without hitting a break statement.

Think of the else block as a "No Breaks Encountered" block. If the loop completes its entire run normally, the else code executes.

If a break is triggered, Python exits the loop instantly and skips the else block entirely.
'''

for i in range(1,11):
    if i == 15 or i == 100:
        break
    print(i)
else: # if break works else will not be exceuted, and vice versa for break
    print("no break was encountered")
```

```python
# reverse in loop 
# Use for i in range(...) when the position or order matters (like skipping steps or going backward)
# Because Python naturally reads strings from left to right, your index loop forces it to look at the string in reverse by starting at the very last index position and counting down to 0.

a= 'python'
rev = " "
for i in range(len(a)-1, -1,-1): # Index-Based (for i in range(...))
    rev = rev + (a[i])
print(f"reverse of {a} is {rev}")
     
```

```python
# count letters, digits, n symbils in a string
''' 

'''
# Use for i in a when you just need to look at every item inside a collection from start to finish

a = "2DELJ3mdkk94#$92mckm!!"
char=0
num=0
schar =0
n = len(a)
for i in a: # Element-Based (for i in a)
    if i.isdigit():
        num += 1
    elif i.isalpha():
        char += 1
    else :
        schar += 1
print(f"total is {n}, letters are {char}, numbers are {num} and special characaters are {schar} and thus again total is {int(num) + int(char)+int(schar)}")
    
```

```python
# ascii values or unicodes
# numbers can be from 0-9 only which range from 48 - 57
# alphabtes a-z -> 97 - 122
# Alphabets A-Z -> 65 - 90
# sChar - 


a = "2DELJ38r4nciu$%^29dkk94#$92mckm!!"
char=0
num=0
schar =0
n = len(a)
for i in a: # Element-Based (for i in a)
    if (ord(i) >= 65 and ord(i) <= 90) or (ord(i) > 96 and ord(i) <123 ):
        char += 1
    elif ord(i) >= 48 and ord(i) <= 57:
        num += 1
    else :
        schar += 1
print(f"total is {n}, letters are {char}, numbers are {num} and special characaters are {schar} and thus again total is {int(num) + int(char)+int(schar)}")
```
