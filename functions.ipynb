```python
''' 
int main() {
    add(2, 3);        // compiler: "what is add? never heard of it"
}

int add(int a, int b) {
    return a + b;
}

----
When the compiler reaches add(2, 3), it hasnot seen add yet. It doesnot know the return type or what parameters it takes, so it cannot check the call is correct.

A declaration fixes this. It tells the compiler: “this function exists, here is its return type and parameters, the full code comes later.”

int add(int a, int b);    // declaration

int main() {
    add(2, 3);            // compiler now knows what add is
}

int add(int a, int b) {   // definition
    return a + b;
}
------
Why it is also useful:

Multiple files: a function defined in one file can be used in another. Declarations (usually in .h header files) tell each file what exists.

Type checking: C is statically typed, so the compiler checks that you pass the right types and use the return value correctly.

Order freedom: you can put main() at the top and the definitions below.
'''
print("yup")
```

```python
# Parameters & Arguments 
# Parameters are the variable names in the function definition. They’re 'placeholders'
# Arguments are the 'actual' values you pass in the function call.

def addition(a,b): # parameters
    
    '''Because parameters are names only and They are placeholders waiting to receive values, so Python doesn’t allow running code like input() there. Python expects just a variable name in that spot, not a function call. '''
    
    print(a+b)
    
addition(12,4)

#addition(int(input("a")),int(input("b"))) #arguments

# Parameter = Placeholder (in the definition)
# Argument = Actual value (in the call)
```

```python
# Types of Arguments - 

# positional arguments : Values go to parameters in order.
def intro(name, age):
    print(name, age)

intro("Sam", 20)     # name="Sam", age=20
intro(20, "Sam")     # wrong order, wrong meaning

# Defualt arguments : The parameter has a default value, used if you don’t pass one.
def addition(a,b,c = 12): # once a default value is set all the next values shld be given defualt values tooo. but this also works (a,b,d,c =0)
    print(a+b+c)
addition(1,2)
addition(1,2,3)

# keyword arguments : You name the parameter, so order doesn’t matter.

def multi(a,b):
    print(a*b)

multi(b=2,a=8) # once a default value is set all the next values shld be given defualt values tooo

multi(20, b= 40)
```
