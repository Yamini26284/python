```python
'''This statement will print True.

print(12>5 or 4>9 and 12 ==12 and 43 !=98)

Step 1: Evaluate the Comparisons First

Comparison operators (>, ==, !=) have a higher precedence than logical operators (and, or). Python evaluates them first from left to right:
• 12 > 5 - True
• 4 > 9 - False
• 12 == 12- True
• 43 != 98 - True

The expression now looks like this:
True or False and True and True

Step 2: Group by Operator Precedence

Because and has a higher precedence than or, Python groups all the and conditions together first:
True or (False and True and True)

Step 3: Left-to-Right Short-Circuiting

Now, Python reads the expression from left to right:
1. It sees the first value is True.
2. It sees the next operator is or.
3. Short-circuit occurs: Because the left side of the or is already True, the entire expression is guaranteed to be True.
Python completely ignores and skips evaluating the rest of the line (False and True and True). It instantly stops and prints True.
'''
print("yup!!")
```

1. not has the highest precedence followed by and , or.  
precendence can be overriden with the help of paranthesis  

2. While and has higher conceptual priority for grouping, Python evaluates expressions from left to right and uses short-circuiting.  

3. For an or operator, if the left side is already True, Python knows the entire statement must be True. It will completely skip evaluating the right side to save time 
 
4. Similarly for and if left side is already false, skip evaluavting the right side to save time.
